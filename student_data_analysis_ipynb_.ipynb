```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    'Student': ['Aman', 'Rahul', 'Priya', 'Neha', 'Riya',
                'Karan', 'Anjali', 'Vikas', 'Simran', 'Arjun'],
    'Maths': [85, 70, 95, 60, 88, 75, 92, 65, 90, 80],
    'Science': [90, 75, 92, 65, 85, 80, 95, 70, 88, 82],
    'English': [80, 85, 90, 70, 92, 78, 88, 75, 95, 85],
    'Attendance': [95, 80, 98, 70, 90, 85, 96, 75, 92, 88],
    'Study_Hours': [4, 2, 6, 1, 5, 3, 6, 2, 5, 4]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head()
```

```python
print("Dataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Marks:")
print(df[['Maths', 'Science', 'English']].mean())
```

```python
print("Statistical Summary:")
print(df.describe())

print("\nAverage Marks:")
print(df[['Maths', 'Science', 'English']].mean())

print("\nHighest Marks:")
print(df[['Maths', 'Science', 'English']].max())

print("\nLowest Marks:")
print(df[['Maths', 'Science', 'English']].min())
```

```python
df['Total_Marks'] = df[['Maths', 'Science', 'English']].sum(axis=1)

df['Average_Marks'] = df[['Maths', 'Science', 'English']].mean(axis=1)

print(df[['Student', 'Total_Marks', 'Average_Marks']])
```

```python
subject_avg = df[['Maths', 'Science', 'English']].mean()

plt.figure(figsize=(8,5))
subject_avg.plot(kind='bar')

plt.title('Subject-wise Average Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.show()
```

```python
df.plot(x='Student',
        y=['Maths', 'Science', 'English'],
        kind='bar',
        figsize=(12,6))

plt.title('Student-wise Academic Performance')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df['Student'], df['Attendance'])

plt.title('Student Attendance Analysis')
plt.xlabel('Students')
plt.ylabel('Attendance (%)')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Study_Hours'], df['Average_Marks'])

plt.title('Study Hours vs Average Marks')
plt.xlabel('Study Hours')
plt.ylabel('Average Marks')
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[['Maths', 'Science', 'English',
        'Attendance', 'Study_Hours']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Student Data Correlation Heatmap')
plt.show()
```

```python
topper = df.loc[df['Total_Marks'].idxmax()]

print("Class Topper:", topper['Student'])
print("Total Marks:", topper['Total_Marks'])
```

```python
weak_students = df[df['Average_Marks'] < 40]

print("Students needing improvement:")
print(weak_students[['Student', 'Average_Marks']])
```

```python
print("Average Attendance:",
      df['Attendance'].mean())

low_attendance = df[df['Attendance'] < 75]

print("\nStudents with low attendance:")
print(low_attendance[['Student', 'Attendance']])
```

```python
print("====== STUDENT DATA ANALYSIS REPORT ======")

print("Total Students:", len(df))

print("Class Average:",
      round(df['Average_Marks'].mean(), 2))

print("Highest Total Marks:",
      df['Total_Marks'].max())

print("Lowest Total Marks:",
      df['Total_Marks'].min())

print("Average Attendance:",
      round(df['Attendance'].mean(), 2))

print("==========================================")
```
